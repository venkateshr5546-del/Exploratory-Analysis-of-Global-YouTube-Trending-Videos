```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

print("Libraries imported successfully!")
```

```python
uploaded = files.upload()

file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Raw dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# Display the first five rows
display(df.head())

# Display the last five rows
display(df.tail())

# Display column names
print("Column Names:")
print(df.columns.tolist())

# Dataset information
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python
# Create a copy of the raw dataset
clean_df = df.copy()

# Remove duplicate rows
clean_df = clean_df.drop_duplicates()

# Remove unnecessary spaces from column names
clean_df.columns = clean_df.columns.str.strip()

# Remove extra spaces from text columns
for col in clean_df.select_dtypes(include=["object"]).columns:
    clean_df[col] = clean_df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Convert numerical columns
numeric_columns = [
    "views",
    "likes",
    "comment_count",
    "category_id"
]

for col in numeric_columns:
    if col in clean_df.columns:
        clean_df[col] = pd.to_numeric(
            clean_df[col], errors="coerce"
        )

# Convert date columns
date_columns = ["publish_time", "trending_date"]

for col in date_columns:
    if col in clean_df.columns:
        clean_df[col] = pd.to_datetime(
            clean_df[col], errors="coerce", utc=(col == "publish_time")
        )

# Replace negative engagement values with missing values
engagement_columns = ["views", "likes", "comment_count"]

for col in engagement_columns:
    if col in clean_df.columns:
        clean_df.loc[clean_df[col] < 0, col] = np.nan

# Remove rows without essential identifiers
if "video_id" in clean_df.columns:
    clean_df = clean_df.dropna(subset=["video_id"])
    clean_df = clean_df[
        clean_df["video_id"].astype(str).str.strip() != ""
    ]

# Remove rows without video titles
if "title" in clean_df.columns:
    clean_df = clean_df.dropna(subset=["title"])
    clean_df = clean_df[
        clean_df["title"].astype(str).str.strip() != ""
    ]

# Save the cleaned dataset
clean_df.to_csv("youtube_trending_cleaned.csv", index=False)

print("Dataset cleaning completed!")
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", clean_df.shape)

display(clean_df.head())
```

```python

# Save the original dataset
df.to_csv("youtube_trending_raw.csv", index=False)

# Save the cleaned dataset
clean_df.to_csv("youtube_trending_cleaned.csv", index=False)

# Download raw dataset
files.download("youtube_trending_raw.csv")

# Download cleaned dataset
files.download("youtube_trending_cleaned.csv")
```

```python
print("CLEANED DATASET SUMMARY")
print("=" * 40)

print("Total Rows:", len(clean_df))
print("Total Columns:", len(clean_df.columns))

print("\nMissing Values:")
print(clean_df.isnull().sum())

print("\nDuplicate Rows:", clean_df.duplicated().sum())

print("\nData Types:")
print(clean_df.dtypes)

print("\nNumerical Summary:")
display(clean_df.describe().T)
```

```python
if {"title", "views"}.issubset(clean_df.columns):

    top_videos = (
        clean_df.dropna(subset=["views"])
        .nlargest(10, "views")
    )

    plt.figure(figsize=(12, 6))

    sns.barplot(
        data=top_videos,
        x="views",
        y="title"
    )

    plt.title("Top 10 YouTube Trending Videos by Views")
    plt.xlabel("Number of Views")
    plt.ylabel("Video Title")
    plt.tight_layout()
    plt.show()

else:
    print("Required columns 'title' and 'views' are missing.")
```

```python
engagement_columns = [
    col for col in ["views", "likes", "comment_count"]
    if col in clean_df.columns
]

if len(engagement_columns) >= 2:

    plt.figure(figsize=(8, 6))

    sns.heatmap(
        clean_df[engagement_columns].corr(),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Between YouTube Engagement Metrics")
    plt.tight_layout()
    plt.show()

else:
    print("Insufficient engagement columns for correlation analysis.")
```

```python
if "country" in clean_df.columns:

    country_counts = clean_df["country"].value_counts()

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=country_counts.index,
        y=country_counts.values
    )

    plt.title("Number of Trending Video Records by Country")
    plt.xlabel("Country")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    display(country_counts.rename("record_count").to_frame())

else:
    print(
        "No country column found. "
        "Combine country-specific datasets and add a country column first."
    )
```
